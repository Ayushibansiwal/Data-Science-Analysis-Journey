#### Oops Problem Practice : Product store

```python
class Products :
    count = 0 

    def __init__(self, name, price):
        self.name = name 
        self.price = price 
        Products.count += 1

    def calc_discount(self,discount):
        print(f'The price of {self.name} after {discount}% discount is {self.price - (discount * self.price) / 100}')


    @classmethod
    def get_total_objects(cls):
        print(f'Total objects are : {cls.count}')

P1 = Products("Laptop",80_000)
P2 = Products("Phone", 50_000)
P3 = Products("Camera", 30_000)

Products.get_total_objects()

P1.calc_discount(20)
```

```python
# Encapsulation : wrapping data and functions into single unit 

#data hiding -> public (inside class and outside : everywhere), protected (class + subclasses) , private (inside the class access only)


# byDefault attributes are public
# _ => protected 
# __ => private 

# data mangling : can't access data directly but indirectly though getters and setters

class BankAccount: 
    def __init__(self, name, balance):
        self.name = name  #public
        self.__balance = balance #protected

    def get_balance(self):
        return self.__balance

    def set_balance(self, newBalance):
        self.__balance = newBalance
        
 

acc1 = BankAccount("Ava", 80_000)
acc1.set_balance(1_00_000)
print(acc1.name, acc1.get_balance())

print(acc1._BankAccount__balance)
        

```

```python
# Inheritance  : Reusing attr and methods from a patent(base) class


class Employee :  #parent  or base class
    start_time = "10AM"
    end_time = "6PM"

class AdminStaff(Employee):
    def __init__(self, role):
        self.role = role

class Accountant(AdminStaff):
    def __init__(self,salary, role):
        super().__init__(role)
        self.salary = salary 



acc1 = Accountant(70_000, "CA")
print(acc1.role,acc1.salary, acc1.start_time, acc1.end_time)

# Super is used to call the constructor of parent class
```

```python
class Teacher:
    def __init__(self, salary, subject):
        self.salary = salary 
        self.subject = subject

class Student :
    def __init__(self, cgpa):
        self.cgpa = cgpa

class TA (Teacher, Student):
    def __init__(self, salary,subject,  cgpa, name):
        super().__init__(salary, subject)
        Student.__init__(self, cgpa)
        self.name = name

TA1 = TA(40_000,"Computerscience",8.79, "Alex")
print(TA1.name, TA1.subject, TA1.salary,TA1.cgpa)


```

```python
# Abstraction : Hding internal details and showing only essential featuees
# abstract classes -< blue print for ther classes 
# present in abc module

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Lion(Animal):
    def make_sound(self):
        print("Roar")

class Cow(Animal):
    def make_sound(self):
        print("Moo!")

L1 = Lion()
L1.make_sound()

C1 = Cow()
C1.make_sound()
```
