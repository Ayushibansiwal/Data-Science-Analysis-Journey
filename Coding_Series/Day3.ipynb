### Constructor 

python allows only one contructor 

```python
class Student :
    def __init__(self,name, cgpa):
        self.name = name
        self.cgpa = cgpa

    def get_cgpa(self):
        return self.cgpa

stu1 = Student("alex",9.7)
stu2 = Student("bob",8.4)

print(stu1.name)
print(stu1.get_cgpa())
    
```

```python
# Attributes : class attributes and instrance attributes 

class Student :
    college_name =  "ABC College"
    PI = 3.14
    def __init__(self, name, cgpa):
        self.name = name 
        self.cgpa = cgpa
        self.PI = 3.14

    
stu1 = Student("Eva",9.2)
stu2 = Student("Drake",9.4)


print(stu1.name)
print(stu1.college_name)
print(Student.college_name) 

print(stu1.PI)
print(Student.PI)
# if class and instance have attribute with same name -< instance have higher priority



```

```python
# Methods 

class Laptop :
    storage_type = "SSD"

    def __init__(self, ram , storage ):
        self.ram = ram 
        self.storage = storage 

    @classmethod
    def get_storage_type(cls):
        print(f'Storage type is {cls.storage_type}')


    @staticmethod
    def get_discount(price, discount):
        print(f'price after discount: {price - discount*price /100}')


    def get_info(self):
        print(f'Laptop has {self.ram}, {self.storage}, {self.storage_type}')

l1 = Laptop("16GB" , "512GB")
l2 = Laptop("8GB" , "256GB")

l1.get_info()
Laptop.get_storage_type()
l1.get_discount(40_000,10)


# instance 
#1. First param -> self
#2. It can access the class and instance attributes
```

#### class methods 
#1. First param -> cls (refer to the Class)
#2. access class Attribute 
#3. classMethod -> is a class decorator : convert instance method in class methods
#4. ojects can call instance methods as well as class methods

#### Static Methods 

#1. no compulsary param | no self , cls param 
#2. can't access class and object's attributes 
#3. @staticmethod -> convert into static method 
#4. do not belong to class and instnace !, they are stand alone functions
