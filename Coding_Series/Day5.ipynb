#### Exception Handling 
try, except, else, finally 

```python

try:
    x = int(input("enter a number : "))
    ans = 10/x
except ZeroDivisionError:
    print("divide by Zero is not allowed.")
except ValueError:
    print("Invalid input.")
else:
    print(f'ans = {ans}')
finally :
    print("i always run")
```
