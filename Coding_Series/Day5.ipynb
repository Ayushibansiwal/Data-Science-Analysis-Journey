#### Exception Handling 
try, except, else, finally 

```python

try:
    x = int(input("enter a number : "))
    ans = 10/x
except ZeroDivisionError:
    print("divide by Zero is not allowed.")
except ValueError:
    print("Invalid input.")
else:
    print(f'ans = {ans}')
finally :
    print("i always run")
```

#### List Comprehensions 

```python
# Output for items in iterable if conditions

# square of numbers
# lst = []
# for i in range(6) :
#     lst.append(i*i)
# print(lst)


lst  =  [i*i for i in range(6)]
print(lst)

```

```python
# sqaure of only odd numbers
lst  = [i*i  for i in range(6)  if i % 2 != 0]
print(lst)
```

```python
lst = [-2,-4,3, 5 , 2, -1]

lst = [0 if i < 0 else i for i in lst ]
print(lst)
```

```python
word = ["Hello" , "Python", "World"]
word = [i.upper() for i in word]
print(word)
```
