Anaconda -> Command line utility -> called conda 


1. conda -> package and env. management system
2. conda -> works to manage different types of virtual environment 


commands to be write on anaconda prompt

commands:
1. activate            Activate a conda environment.
2. deactivate          Deactivate the current active conda
                        environment.
3. create              Create a new conda environment from
                        a list of specified packages.
4. env                 Create and manage conda
                        environments.
5. list                List installed packages in a conda
                        environment.
