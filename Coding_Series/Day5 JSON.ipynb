### JSON Module


Json -> javascript object Notation 
most comman format in which developers stores their data.

json like dictory : stores data in key:val pairs

```python
#   arrays in json  -> list in python
#   objects in json -> dictionary in python
#   string in json -> str in python
#   null in json -> None in python


#json module 
# import json  

#json Methods 
# 1. json.loads()  # s -> string
# 2. json.dumps()  # s -> string
'''
json String -> python object :loads
json String <- python object :dumps
'''
#__________________________

# 3. json.load()
# 4. json.dump()
```

```python
import json 
json_str = '{ "name" : "ava", "age" : 19 , "isStudent" : true}' 

py_obt = json.loads(json_str)
print(py_obt, type(py_obt))
```

```python
py_obt = {
    "name" : "alex",
    "isStudent" : False,
    "Subject" : "Python"
}

json_str = json.dumps(py_obt)
print(type(json_str) , json_str)
```

```python
# json.load() -> read 
# json.dump() -> write 

d = {
    "name" : "bob",
    "age" : 19, 
    "isTeacher" : False
}

# with open("data.json", "r") as f:
#     py_obt = json.load(f)
#     print(py_obt)

with open("data.json", "w") as f:
    json.dump(d,f, indent=4, sort_keys=True)
```
