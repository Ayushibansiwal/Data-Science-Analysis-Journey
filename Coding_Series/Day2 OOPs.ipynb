### OOPs 
object oriented programming

function : readability | reusability


class: A class is a blueprint or template used to create objects. It defines the properties (data) and methods (functions) of an object.

object : An object is an instance of a class. It represents a real-world entity and has its own properties and methods.


```python
class Student :
    subject = "Python"

s1 = Student()
print(s1.subject)
```

```python
'''
class ->    Propertires  |  Behaviours
            Attribute   |   Methods
'''

# constructor  : to initialise our objects
# it is a special method that get called everytime a object is created (if we don't create constructor , python run default constructor)

# self : stores the reference to the current object 

class Student :
    def __init__(self, name, cgpa):
        self.name = name
        self.cgpa = cgpa

    def get_cgpa(self):
        return self.cgpa

student1 = Student("alex",9.7)
student2 = Student("bob",8.6)
student3 = Student("casey",7.9)

print(student1.name, student1.get_cgpa())
print(student2.name, student2.get_cgpa())
print(student3.name, student3.get_cgpa())

```

```python
# types of constructor 
# default | parameterised

# only one param : self  -> default
# more param :  parameterized

# python doesn't allow multiple constructor 


```
