### Conditional Statement

Match case : alternate for if-elif-else

```python
color = input("Enter a color")

match color : 
    case "Green" :
        print("Go")
    case "Yellow" : 
        print("Wait")
    case "Red" :
        print("Stop")
    case _ :
        print("Wrong color")
```

### Loops 

while, for 

```python
count = 0 
while count < 5 :
    print(f'{count+1}. hello world')
    count += 1
```

```python
# print reverse counting from 10 10 1 
i = 10 
while i >= 1 :
    print(i)
    i -=1
```

```python
# multiplication table of any number n 
n = int(input("Enter a num : "))
for i in range(1, 11):
    print(f' {n} * {i}  = {n * i}')
```

#### break  | continue 

```python
# break on multiple of 3 
for i in range(1, 11):
    if( i % 3 == 0):
        # break
        continue
    else :
        print(i)

# print all the odd numbers from 1 to 10 
for i in range(1, 11):
    if( i % 2 != 0):
        print(i)
    else : continue
```

```python
# vowel count
sentence = input("enter : ")
vowel_count = 0 
for i in sentence :
    if (i in "AEIOUaeiou"):
        vowel_count +=1 

print(f' count : {vowel_count}')
```

```python
# sum of first natural numbers
n = 2 
sum = 0 
for i in range(1, n+1): 
    sum += i
print(sum)
```

### Functions 

```python
# average of three numbers
num1 = 10 
num2 = 20 
num3 = 30 

def avg(a,b,c):
    return (a+b+c)/3

print(f'average is {avg(num1, num2,num3)}')
```

```python
# Dictionary : key:value pairs | mutable | unordered
dict = {
    "name" : "ava",
    "rollno" : 19,
    "age" : 19 ,
    "city" : "new delhi",
    "class" : 12,
    "subjects" : ["maths", "chemistry", "physics"],
    23 : "jersey number"
}

# print(dict)

# print(dict["name"])

# mutable
dict["age"]  = 20
print(dict["age"])
```

```python
# dict methods

# return all keys 
dict.keys()

# return all values 
dict.values()

# return all key:value pairs
dict.items()

# returns val acc to key 
dict.get("name") # do  not give error if key doesn't exist
dict.get("subjects")

# adds new item to dict 
dict.update({"country" : "india"})

# remove  by key
# dict.pop("age")

# remove 
dict.popitem()
print(dict)
```
