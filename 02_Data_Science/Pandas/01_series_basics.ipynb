#### **Pandas**
Data analysis and manipulation tool 
 - build on top of NumPy

```python
import pandas as pd
```

1. **series ->  1d labelled array**
2. **dataframe -> 2d labelled array (row, cloumn)**

```python
info = {
    "Name" : ["alex","bob","casey","drake","eva"],
    "CGPA" : [9.7, 8.3, 8.9, 7.9, 9.2]
}

df = pd.DataFrame(info)
print(df)
```

```python
# series (1D labelled array)
series = pd.Series(["apple","grapes","mango","cherry","guava"])
print(series)

# index based access
print(series[3])
print(series[2])
```

```python
age = [18,19,20,21,22]

s = pd.Series(age, index=["alex","bob","casey","drake","eva"])

print(s)

# labell (named index) based access
print(s["casey"])
print(s["alex"])

# all labels 
print(s.index)
```

##### **Series Properties**
1. homogenoues 
2. vectorised operations 
3. handle misssing values with NaN
4. mustble values, immutable size

```python
# homogenous
marks = [18,19,20,21,"abc"]
s = pd.Series(marks, index=["alex","bob","casey","drake","eva"])
print(s) 

# it will make the dtype common for all -> object
```

```python
# vectorise operations 
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])

print(s1+s2)
```

```python
# Handle missing value with NaN
s = pd.DataFrame([1,2,3,4, None])
print(s)
```

```python
# Mutable values and immutable size 
marks = [18,19,20,21,20]
s = pd.Series(marks)

s[0] = 250
print(s)

changed_s1 = s1.drop(0)
print("_______")
print(s)
print("_______")
print(changed_s1)
```
