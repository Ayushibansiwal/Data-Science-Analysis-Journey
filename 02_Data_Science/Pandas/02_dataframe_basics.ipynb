#### **DataFrame (2D Labelled array)**

```python
import pandas as pd
```

```python
# dataframe using dictionary 
info = {
    "Name" : ["alex","bob","casey","drake","eva"],
    "Age" : [19,20,19,19,20],
    "CGPA" : [9.2,9.3,8.9,9.1,8.7]
}

df = pd.DataFrame(info)
print(df)

# get row label
print(df.index)
print(df.columns)
```

```python
# dataframe using lists of lists
df = pd.DataFrame([["alex",19],["bob",17],["casey",20]], columns=["Name","Age"])
print(df)
```

```python
# dataframe using numpy
import numpy as np

arr = np.array([["apple","grapes","cherry","guava"],["red","green","red","green"]])
df = pd.DataFrame(arr.T, columns=["Fruits","Color"])
print(df)

# arr.T -> transpose
```
