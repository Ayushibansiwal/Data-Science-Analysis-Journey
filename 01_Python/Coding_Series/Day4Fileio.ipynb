```python
# Writing in file : w -> over write the entire file

file = open("Data.txt", "w") # file path |  mode
file.write("Text to overwrite \nthe complete data.")
file.close()
```

Modes 
1. r: Reading (by defauLT)
2. w: Writing, truncates file first
3. x: creates new and open for writing 
4. a: writing , appends at the end 
5. b: binary mode 
6. t: text mode(by default)
7. +: open disk file for update(r & w)

```python
file = open("Data.txt", "a") # file path |  mode
file.write("\nnew text being append in the file.")
file.close() 
```

combined modes 
1. rt | wt
2. rb | wb
3. r+ w+ a+

```python
file = open("text.txt","r+") # read : pointer in the beginning 
file.write("123") #first wirte :  A b : will get replaced by 123
print(file.read()) #then reading starts from ook is like a magic door
file.close()
```

```python
file = open("text.txt","a+") # append : pointer in the end of the file 
file.write("123") #first wirte :  123 in the end
print(file.read()) #then reading starts from end but nothing to read
file.close()
```

```python
file = open("text.txt","w+") # write : first clear the whole content
file.write("123") #then write , pointer in the beginning  
print(file.read()) #then reading starts : pointer at the end : nothing to read
file.close()
```

```python
with open("text.txt","r") as f:
    print(f.read())


# with automatically closes the file ! 
```
