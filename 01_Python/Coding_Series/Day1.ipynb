Operators : 

Arithmetic :  + - / // % *  **

Relational/Comparison :  >  <  >= <=  == !=

Assignment  : =  += -= **= /=

logical : and or not

membership : in

Operator precedence (priority order) : 
Priority	Operator	    Meaning
1	        ()	            Parentheses
2	        **	            Exponentiation
3	        +x, -x	        Unary plus/minus
4	        *, /, //, %	    Multiplication, division, etc.
5	        +, -	        Addition, subtraction
6	        ==, !=, >, <	Comparisons
7	        not	            Logical NOT
8	        and	            Logical AND
9	        or	            Logical OR

Associativity : determines the order in which operators of the same precedence are evaluated in an expression.

Associativity	       Operators
Left to Right	       +, -, *, /, //, %
Right to Left	       **

#### Conditional Statements

```python
age = 90 
if age >= 18 : 
    print("vote") 
else:
    print("no vote")
```

```python
traffic_light = "yellow"

if (traffic_light == "red"):
    print("stop")
elif (traffic_light == "yellow"):
    print("Wait")
else:
    print("go")
```

```python
age = 20 
if age < 13 : 
    print("child")
elif  age >= 13 and age < 18:
    print("teenager")
elif  age >= 18 :
    print("adult")
```

```python
# check multiple of 5 
num = 19
if(num % 5 == 0):
    print("yes")
else: print("no")
```

```python
# check odd and even 
num = 10
if( num % 2 == 0):
    print("even")
else:
    print("odd") 
```
